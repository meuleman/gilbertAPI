```python
def roiurl_dataset(query_region, datatype, top_regions):
    today = date.today()
    roiurl_name = datatype + '_similarities_region_' + query_region + '_25kb_CT' + today.strftime("%Y%m%d") + '.bed'
    bed_path_name = os.path.join('/home/ctrader/public_html/', roiurl_name)
    np.savetxt(bed_path_name, top_regions, delimiter='\t', fmt='%s')
    if datatype == 'epilogos':
        link_path = 'https://epilogos.altius.org:1234/?application=viewer&sampleSet=vC&mode=single&genome=hg38&model=18&complexity=KL&group=all&roiURL=https%3A%2F%2Fencode%3Acollabor8%40resources.altius.org%2F~ctrader%2F' + roiurl_name
    elif datatype == 'DHS':
        link_path = 'https://index.altius.org:1234/?application=viewer&mode=pd&roiURL=https%3A%2F%2Fencode%3Acollabor8%40resources.altius.org%2F~ctrader%2F' + roiurl_name
    return link_path
```

**Fusion**

```python
num_regions = len(DHS_top_sim)

top_100_regions_split_by_contributions_p = []
top_100_fusion_scores_p = []
winning_dataset_p = []

for i in range(num_regions):
    if i % 10000 == 0:
        print(i)
    DHS_arr = np.zeros(num_regions)
    TF_arr = np.zeros(num_regions)
    epi_arr = np.zeros(num_regions)

    #setting similarity values for each region
    DHS_arr[DHS_top_regions[i]] = DHS_top_sim[i]
    TF_arr[TF_top_regions[i]] = TF_top_sim[i]
    epi_arr[epi_top_regions[i]] = epi_top_sim[i]
    
    #fusion score with priors
    fusion_i = 1 - ((1 - (DHS_priors[i] * DHS_arr)) * 
                    (1 - (TF_priors[i] * TF_arr)) * 
                    (1 - (epi_priors[i] * epi_arr)))
    
    #how many of top 100 are from each metric
    top_100 = np.argsort(-fusion_i)[:100]
    top_100_fusion_scores_p.append(-np.sort(-fusion_i)[:100])
    #which dataset drives similarity
    winning_dataset_p.append(np.argmax([np.mean(DHS_arr[top_100]),
                                        np.mean(TF_arr[top_100]),
                                        np.mean(epi_arr[top_100])]))
    
    all3 = len(top_100[np.isin(top_100, DHS_top_regions[i]) & 
                       np.isin(top_100, TF_top_regions[i]) & 
                       np.isin(top_100, epi_top_regions[i])])
    
    notDHS = len(top_100[~np.isin(top_100, DHS_top_regions[i]) & 
                         np.isin(top_100, TF_top_regions[i]) & 
                         np.isin(top_100, epi_top_regions[i])])
    
    notTF = len(top_100[np.isin(top_100, DHS_top_regions[i]) & 
                        ~np.isin(top_100, TF_top_regions[i]) & 
                        np.isin(top_100, epi_top_regions[i])])
    
    notepi = len(top_100[np.isin(top_100, DHS_top_regions[i]) & 
                         np.isin(top_100, TF_top_regions[i]) & 
                         ~np.isin(top_100, epi_top_regions[i])])
    
    dhsonly = len(top_100[np.isin(top_100, DHS_top_regions[i]) & 
                          ~np.isin(top_100, TF_top_regions[i]) & 
                          ~np.isin(top_100, epi_top_regions[i])])
    
    TFonly = len(top_100[~np.isin(top_100, DHS_top_regions[i]) & 
                         np.isin(top_100, TF_top_regions[i]) & 
                         ~np.isin(top_100, epi_top_regions[i])])
    
    epionly = len(top_100[~np.isin(top_100, DHS_top_regions[i]) & 
                          ~np.isin(top_100, TF_top_regions[i]) & 
                          np.isin(top_100, epi_top_regions[i])])

    top_100_regions_split_by_contributions_p.append([all3, notDHS, notTF, notepi, dhsonly, TFonly, epionly])
top_100_regions_split_by_contributions_p = np.array(top_100_regions_split_by_contributions_p)
top_100_fusion_scores_p = np.array(top_100_fusion_scores_p)
winning_dataset_p = np.array(winning_dataset_p)
```

```python
def roiurl_fusion(query_region, top_regions):
    today = date.today()
    roiurl_name = 'fusion_similarities_region_' + query_region + '_25kb_CT' + today.strftime("%Y%m%d") + '.bed'
    bed_path_name = os.path.join('/home/ctrader/public_html/', roiurl_name)
    np.savetxt(bed_path_name, top_regions, delimiter='\t', fmt='%s')
    link_path_epi = 'https://epilogos.altius.org:1234/?application=viewer&sampleSet=vC&mode=single&genome=hg38&model=18&complexity=KL&group=all&roiURL=https%3A%2F%2Fencode%3Acollabor8%40resources.altius.org%2F~ctrader%2F' + roiurl_name
    link_path_index = 'https://index.altius.org:1234/?application=viewer&mode=pd&roiURL=https%3A%2F%2Fencode%3Acollabor8%40resources.altius.org%2F~ctrader%2F' + roiurl_name
    print(link_path_epi)
    print(link_path_index)
    return link_path_epi, link_path_index
```

```python
num_regions = len(DHS_top_sim)
for i in np.argsort(-top_100_fusion_scores_p[:, 0])[[0, 30000]]:
    icoord = regions[i]
    icoord_print = icoord[0] + ':' + icoord[1] + '-' + icoord[2]
    print(icoord_print)
    
    DHS_arr = np.zeros(num_regions)
    TF_arr = np.zeros(num_regions)
    epi_arr = np.zeros(num_regions)

    #setting similarity values for each region
    DHS_arr[DHS_top_regions[i]] = DHS_top_sim[i]
    TF_arr[TF_top_regions[i]] = TF_top_sim[i]
    epi_arr[epi_top_regions[i]] = epi_top_sim[i]
    
    #fusion score with priors
    fusion_i = 1 - ((1 - (DHS_priors[i] * DHS_arr)) * 
                    (1 - (TF_priors[i] * TF_arr)) * 
                    (1 - (epi_priors[i] * epi_arr)))
    
    #top 99
    top_99 = np.argsort(-fusion_i)[:99]
    #scores
    DHS_scores = np.round(DHS_arr[top_99], 3).astype(str)
    TF_scores = np.round(TF_arr[top_99], 3).astype(str)
    epi_scores = np.round(epi_arr[top_99], 3).astype(str)
    fusion_scores = np.round(fusion_i[top_99], 4).astype(str)
    top_coords = regions[top_99]
    #top regions for roiURL
    slash_arr = np.array([' / '] * 99)
    scores_col = np.char.add(fusion_scores, slash_arr)
    scores_col = np.char.add(scores_col, DHS_scores)
    scores_col = np.char.add(scores_col, slash_arr)
    scores_col = np.char.add(scores_col, TF_scores)
    scores_col = np.char.add(scores_col, slash_arr)
    scores_col = np.char.add(scores_col, epi_scores)
    top_regions = np.hstack((top_coords, scores_col.reshape(-1, 1)))
    
    query_region = np.append(regions[i], ['Fusion / DHS (' + str(np.round(DHS_priors[i], 3)) +
                                           ') / TF (' + str(np.round(TF_priors[i], 3)) +
                                           ') / Epilogos (' + str(np.round(epi_priors[i], 3)) + ')'])
    
    top_regions = np.vstack((query_region, top_regions))
    roiurl_fusion(str(i) + '_' + icoord_print.replace(':', '_').replace('-', '_'), 
                  top_regions)
```
