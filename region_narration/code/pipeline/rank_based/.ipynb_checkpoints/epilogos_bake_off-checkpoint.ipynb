**Functions**

```python
def region_percentile(arr):
    # rank each region
    ranked_regions = rankdata(arr)

    # find the number of regions with 0 values for each component
    comp_num_zero = len(arr) - np.count_nonzero(arr)

    # adjusting/resetting ranks
    ranked_regions = ranked_regions - comp_num_zero
    ranked_regions = np.where(ranked_regions < 0, 0, ranked_regions)
    
    percentile = ranked_regions / np.max(ranked_regions)

    return percentile
```

```python
def print_coords(i):
    c = regions[i]
    return c[0] + ':' + c[1] + '-' + c[2]
```

```python
def roiurl(top_regions, method):
    today = date.today()
    roiurl_name = 'top_EnhWk_regions_' + method + '_25kb_CT' + today.strftime("%Y%m%d") + '.bed'
    bed_path_name = os.path.join('/home/ctrader/public_html/', roiurl_name)
    np.savetxt(bed_path_name, top_regions, delimiter='\t', fmt='%s')
    link_path_epi = 'https://epilogos.altius.org:1234/?application=viewer&sampleSet=vC&mode=single&genome=hg38&model=18&complexity=KL&group=all&roiURL=https%3A%2F%2Fencode%3Acollabor8%40resources.altius.org%2F~ctrader%2F' + roiurl_name
    print(link_path_epi)
    return
```

**Difference between averaging scores across bins and calculating scores for each region with epilogos formula**

```python
exp = 0.4
obs = np.random.random_sample(125) / 2
# obs = np.ones(125) - 0.1
# obs[0] = 0.99
obs
```

```python
# averaging epilogos scores across region
np.mean(obs * np.log2(obs / exp))
```

```python
# calculating epilogos score for a whole region
np.mean(obs) * np.log2(np.mean(obs) / exp)
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10,10))
fig.patch.set_facecolor('white')
fig.subplots_adjust(wspace=0)

axes = axes.ravel()

d = obs
sns.kdeplot(d, bw_adjust=0.1, ax=axes[0])
dmin = np.min(d)
dmax = np.max(d)

d = obs / exp
sns.kdeplot(d, bw_adjust=0.1, c='orange', ax=axes[0])
dmin = min(dmin, np.min(d))
dmax = max(dmax, np.max(d))

d = np.log2(obs / exp)
sns.kdeplot(d, bw_adjust=0.1, c='green', ax=axes[0])
dmin = min(dmin, np.min(d))
dmax = max(dmax, np.max(d))

d = np.mean(obs * np.log2(obs / exp))
axes[0].axvline(x=d, c='black')
dmin = min(dmin, d)
dmax = max(dmax, np.max(d))
# axes[0].set_xlim([dmin, dmax])

axes[0].legend(['obs', 'obs / exp', 'np.log2(obs / exp)', 'np.mean(obs * np.log2(obs / exp))'])



d = np.mean(obs)
axes[1].axvline(d)
dmin = min(dmin, np.min(d))
dmax = max(dmax, np.max(d))

d = np.mean(obs) / exp
axes[1].axvline(d, c='orange')
dmin = min(dmin, np.min(d))
dmax = max(dmax, np.max(d))

d = np.log2(np.mean(obs) / exp)
axes[1].axvline(d, c='green')
dmin = min(dmin, np.min(d))
dmax = max(dmax, np.max(d))

d = np.mean(obs) * np.log2(np.mean(obs) / exp)
axes[1].axvline(x=d, c='black')

axes[1].legend(['np.mean(obs)', 'np.mean(obs) / exp', 'np.log2(np.mean(obs) / exp)', 'np.mean(obs) * np.log2(np.mean(obs) / exp)'])

axes[0].set_xlim([dmin-1, dmax+1])
axes[1].set_xlim([dmin-1, dmax+1])
axes[0].set_title('Method 1 = %s' %(np.round(np.mean(obs * np.log2(obs / exp)), 4)))
axes[1].set_title('Method 2 = %s' %(np.round(np.mean(obs) * np.log2(np.mean(obs) / exp), 4)))
```
