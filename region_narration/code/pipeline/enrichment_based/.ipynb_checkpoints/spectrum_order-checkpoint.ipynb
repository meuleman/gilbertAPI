```python
#kde plot (based on geneXgeneset jaccard clustering)

roi = 6434

#finding the most enriched gene sets to show in colorbar
hdf5_file_path = '/home/ctrader/region_similarity/data/regions_25kb_CT20220912.genesethallmarkc1c2freqooe.gsregionbon.sig.hdf5'
hdf5_file = h5py.File(hdf5_file_path, 'r')
gene_set_sig_log2ooe = hdf5_file['sig_gene_set_roi_log2ooe'][roi]
hdf5_file.close()

gene_set_order_path = '/home/ctrader/region_similarity/data/hallmarkc1c2genesetorder.geneXgs.jaccard.average.optimalorder.CT20220912.npz'
gene_set_order = np.load(gene_set_order_path)['gene_set_order']

hdf5_file_path = '/home/ctrader/region_similarity/data/regions_25kb_CT20220912.genesethallmarkc1c2freqooe.hdf5'
hdf5_file = h5py.File(hdf5_file_path, 'r')
gene_sets = np.char.decode(hdf5_file['gene_sets'][:], 'utf-8')
hdf5_file.close()

#everything in clustered order - alphabetical order, then cluster order
gene_set_sig_log2ooe = gene_set_sig_log2ooe[np.argsort(gene_sets)][gene_set_order]
gene_sets = np.sort(gene_sets)[gene_set_order]

colors = cm.get_cmap('nipy_spectral', len(gene_set_order))

#max over a given window size
window_size = 20
gene_set_sig_log2ooe_windowed = np.array([])
for i in range(len(gene_set_sig_log2ooe)):
    gene_set_sig_log2ooe_windowed = np.append(gene_set_sig_log2ooe_windowed, 
                                              np.max(gene_set_sig_log2ooe[max(i-int(window_size/2), 0):
                                                                          min(i+int(window_size/2), 
                                                                              len(gene_set_sig_log2ooe))]))
#smoothing with mean
gene_set_sig_log2ooe_windowed_smoothed = np.array([])
for i in range(len(gene_set_sig_log2ooe_windowed)):
    gene_set_sig_log2ooe_windowed_smoothed = np.append(gene_set_sig_log2ooe_windowed_smoothed, 
                                                       np.mean(gene_set_sig_log2ooe_windowed[max(i-int(window_size/2), 0):
                                                                                             min(i+int(window_size/2), 
                                                                                                 len(gene_set_sig_log2ooe_windowed))]))


#find the enrichment values that are greater than 0
gene_sets_inds = np.where(gene_set_sig_log2ooe!=0)[0]
gene_sets_vals = gene_set_sig_log2ooe[gene_sets_inds]

cluster_inds_sorted = gene_sets_inds
gene_sets_labels_ordered = gene_sets[gene_sets_inds]
gene_sets_vals_ordered = gene_sets_vals

gene_sets_to_label = np.array([])
cluster_inds_to_label = np.array([])
vals_to_label = np.array([])
for cluster_ind, label, val in zip(cluster_inds_sorted, gene_sets_labels_ordered, gene_sets_vals_ordered):
    label = ' '.join(label.split('_'))
    if len(vals_to_label) == 0:
        gene_sets_to_label = np.append(gene_sets_to_label, label)
        cluster_inds_to_label = np.append(cluster_inds_to_label, cluster_ind)
        vals_to_label = np.append(vals_to_label, val)
    elif ((cluster_ind - cluster_inds_to_label[-1] <= 100) & (vals_to_label[-1] < val)):
        gene_sets_to_label[-1] = label
        cluster_inds_to_label[-1] = cluster_ind
        vals_to_label[-1] = val
    elif cluster_ind - cluster_inds_to_label[-1] > 100:
        gene_sets_to_label = np.append(gene_sets_to_label, label)
        cluster_inds_to_label = np.append(cluster_inds_to_label, cluster_ind)
        vals_to_label = np.append(vals_to_label, val)
        
#trimming terms
for i in range(len(gene_sets_to_label)):
    gene_sets_to_label[i] = gene_sets_to_label[i].replace('WP ', '').replace('HALLMARK ', '').replace('KEGG ', '').replace('REACTOME ', '').replace('BIOCARTA ', '')
    
#plot
fig, axes = plt.subplots(nrows=2, gridspec_kw={'height_ratios': [4, 1]}, figsize=(12, 2))
fig.patch.set_facecolor('white')
axes = axes.ravel()
plt.subplots_adjust(hspace=0)

axes[0].fill_between(range(len(gene_set_sig_log2ooe_windowed_smoothed)), 
                     gene_set_sig_log2ooe_windowed_smoothed, color='k')

matplotlib.colorbar.ColorbarBase(axes[1], cmap=colors, orientation='horizontal').set_ticks(cluster_inds_to_label/len(gene_sets))
axes[0].set_xlim([0, len(gene_sets)])
axes[0].set_xticks([])
axes[0].set_ylabel('Windowed Enrichment')
axes[0].set_ylim([0, np.max(gene_set_sig_log2ooe_windowed)*1.1])
axes[1].set_xticklabels(gene_sets_to_label, rotation=45, ha='right')

for xlabel, size in zip(axes[1].xaxis.get_ticklabels(), vals_to_label):
    xlabel.set_fontsize(size*2.5)

    
f = open(regions_path, 'r')
regions = f.readlines()
f.close()
regions = regions[roi].strip().split('\t')
genome_coords = regions[0] + ':' + regions[1] + '-' + regions[2]

title = 'Smoothed Windowed Enrichment of Significant OOE Gene Sets for ROI %s\n(%s) and Top 100 Regions' %(roi, genome_coords)
fig.suptitle(title, y=1.08)
# plt.savefig('/home/ctrader/region_similarity/data/figures/' + title.replace(' ', '_').replace(':', '').replace('(', '').replace(')', '').replace('\n', '_') + '_CT' + today.strftime("%Y%m%d") + '.pdf', 
#             bbox_inches='tight', 
#             dpi=400, 
#             facecolor="#FFFFFF", 
#             edgecolor="#FFFFFF", 
#             transparent=False)
```

```python
from scipy.stats import hypergeom
def fishers_test_greater_OR(table):
    #code modified from https://github.com/scipy/scipy/blob/v1.8.1/scipy/stats/_stats_py.py#L4120-L4372
    #scipy stats fisher's exact test, but so that we can perform multiple tests at once
    c = np.asarray(table, dtype=np.int64)
    
    oddsratio = np.zeros(len(c[0, 0, :]))

    if np.any(c < 0):
        raise ValueError("All values in `table` must be nonnegative.")

    inf_inds = np.where((c[1, 0, :]==0) | (c[0, 1, :]==0))[0]
    non_inf_inds = np.where((c[1, 0, :]>0) & (c[0, 1, :]>0))[0]
    
    oddsratio[non_inf_inds] = c[0, 0, non_inf_inds] * c[1, 1, non_inf_inds] / (c[1, 0, non_inf_inds] * c[0, 1, non_inf_inds])
    oddsratio[inf_inds] = np.array([np.inf]*len(inf_inds))
    
    return oddsratio
```

```python
#could optimize if desired (https://github.com/scipy/scipy/blob/v1.9.1/scipy/stats/_stats_py.py#L4492-L4723)
OR_sim_m = []
genesXgs_bool_T = genesXgs_bool.T
for gs1 in range(len(genesXgs_bool[0]))[:1]:
#     for gs2 in range(len(genesXgs_bool[0])):
    t = time.time()
    gs1_arr = genesXgs_bool[:, gs1]
    print(time.time()-t)
#         gs2_arr = genesXgs_bool[:, gs2]
    a = np.sum((gs1_arr & genesXgs_bool_T), axis=1)
    print(time.time()-t)
    b = np.sum((gs1_arr & ~genesXgs_bool_T), axis=1)
    print(time.time()-t)
    c = np.sum((~gs1_arr & genesXgs_bool_T), axis=1)
    print(time.time()-t)
    d = np.sum((~gs1_arr & ~genesXgs_bool_T), axis=1)
    print(time.time()-t)
    conting = np.array([[a, b], [c, d]])
    print(time.time()-t)
    OR_sim_m.append(fishers_test_greater_OR(conting))
    print(time.time()-t)
```

```python
##Creating GeneXGeneSet matrix

#hallmark, c1, c2 gene sets
genes_total = np.array([])
genesets_total = np.array([])

for file in ['/home/ctrader/region_similarity/data/h.all.v7.5.1.symbols.gmt',
             '/home/ctrader/region_similarity/data/c1.all.v7.5.1.symbols.gmt',
             '/home/ctrader/region_similarity/data/c2.all.v7.5.1.symbols.gmt']: #for each file (hallmark, c1, c2)
    f = open(file, 'r')
    for line in f.readlines(): #for each gene set and associated genes
        line = line.strip().split()
        gs = line[0] #gene set
        genesets_total = np.append(genesets_total, gs)
        genes = line[2:] #genes for geneset
        genes_total = np.append(genes_total, genes)    
    f.close()
            
#sort in alphabetical order
genes_total = np.unique(genes_total) #only 1 of each gene 
genesets_total = np.sort(genesets_total)

#create matrix
genesXgs = np.zeros((len(genes_total), len(genesets_total)), dtype=int)
for file in ['/home/ctrader/region_similarity/data/h.all.v7.5.1.symbols.gmt',
             '/home/ctrader/region_similarity/data/c1.all.v7.5.1.symbols.gmt',
             '/home/ctrader/region_similarity/data/c2.all.v7.5.1.symbols.gmt']: #for each file (hallmark, c1, c2)
    f = open(file, 'r')
    for line in f.readlines(): #for each gene set and associated genes
        line = line.strip().split()
        gs = line[0] #gene set
        genes = line[2:] #genes for geneset
        genesXgs[np.isin(genes_total, genes), np.isin(genesets_total, gs)] = 1 #set value for genes and gs to 1
    f.close()
```

```python
def fishers_test_greater(table):
    #code modified from https://github.com/scipy/scipy/blob/v1.8.1/scipy/stats/_stats_py.py#L4120-L4372
    #scipy stats fisher's exact test, but so that we can perform multiple tests at once
    c = np.asarray(table, dtype=np.int64)

    if np.any(c < 0):
        raise ValueError("All values in `table` must be nonnegative.")

    n1 = c[0, 0, :] + c[0, 1, :]
    n2 = c[1, 0, :] + c[1, 1, :]
    n = c[0, 0, :] + c[1, 0, :]

    pvalue = hypergeom.cdf(c[0, 1, :], n1 + n2, n1, c[0, 1, :] + c[1, 1, :])
    pvalue = np.where(pvalue > 1, 1, pvalue)

    return pvalue
```

```python
def remove_num_terms(term_list):
    out_term_list = []
    for term in term_list:
        add_to_out = True
        for c in term:
            if c.isdigit():
                add_to_out = False
        if add_to_out:
            out_term_list.append(term)
    return np.array(out_term_list)
```

```python
from scipy.stats import hypergeom
def fishers_test_greater(table):
    #code modified from https://github.com/scipy/scipy/blob/v1.8.1/scipy/stats/_stats_py.py#L4120-L4372
    #scipy stats fisher's exact test, but so that we can perform multiple tests at once
    c = np.asarray(table, dtype=np.int64)

    if np.any(c < 0):
        raise ValueError("All values in `table` must be nonnegative.")

    n1 = c[0, 0, :] + c[0, 1, :]
    n2 = c[1, 0, :] + c[1, 1, :]
    n = c[0, 0, :] + c[1, 0, :]

    pvalue = hypergeom.cdf(c[0, 1, :], n1 + n2, n1, c[0, 1, :] + c[1, 1, :])
    pvalue = np.where(pvalue > 1, 1, pvalue)

    return pvalue
```

```python
roi=71705
# roi=12387
region_counts = gene_set_full_counts[roi]
a = region_counts
b = np.array([101]*len(region_counts))
c = c_sum_all
d = np.array([101*len(gene_set_full_counts)]*len(region_counts))
conting = np.array([[a, b],
                    [c, d]])
pvals = fishers_test_greater(conting)
```
