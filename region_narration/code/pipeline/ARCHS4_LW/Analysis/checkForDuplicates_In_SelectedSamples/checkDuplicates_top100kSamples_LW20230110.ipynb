```python
# Read Data
import ast

root = "/Users/lwu/PycharmProjects/NarrativeGenomics/region_narration/code/pipeline/ARCHS4_LW"

with open(root + "/data_results/top_100k_samples_LW20230108.txt", "r") as file:
    # read the file contents
    data = file.read()
    # Use ast.literal_eval to safely evaluate the string as a dictionary
    selected_samples = ast.literal_eval(data)
```

```python
# Unique Samples
print("total selected samples (should be 100k):", len(selected_samples))
total_unique_samples = len(list(set(selected_samples)))
print("total_unique_samples:", total_unique_samples)
```

```python
# Duplicates
from collections import Counter

counts = Counter(selected_samples)
duplicates = {key: value for key, value in counts.items() if value > 1}
print(duplicates)
```

```python
samples_duplicate_2 = sum(val==2 for val in duplicates.values())
print("Samples - Repeated Twice:", samples_duplicate_2)
```

```python
samples_duplicate_3 = sum(val==3 for val in duplicates.values())
print("Samples - Repeated Thrice:", samples_duplicate_3)
```

```python
samples_duplicate_4 = sum(val==4 for val in duplicates.values())
print("Samples - Repeated 4:", samples_duplicate_4)
```

```python
samples_duplicate_5 = sum(val==5 for val in duplicates.values())
print("Samples - Repeated 5:", samples_duplicate_5)
```
