```python
def count_elements(d):
    count = 0
    for k, v in d.items():
        if isinstance(v, list):
            for item in v:
                if isinstance(item, list):
                    count += len(item)
                elif isinstance(item, dict):
                    count += count_elements(item)
                else:
                    count += 1
        elif isinstance(v, dict):
            count += count_elements(v)
    return count




total_count = count_elements(category_dict)
print(total_count)
```
