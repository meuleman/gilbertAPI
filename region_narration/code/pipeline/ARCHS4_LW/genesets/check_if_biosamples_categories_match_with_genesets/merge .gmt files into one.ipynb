```python
import os

def merge_files(filenames, output_file):
    with open(output_file, 'w') as outfile:
        for filename in filenames:
            with open(filename) as infile:
                for line in infile:
                    outfile.write(line)

# List of input files
filenames = ['h.all.v7.5.1.symbols.gmt', 'c1.all.v7.5.1.symbols.gmt', 'c2.all.v7.5.1.symbols.gmt']

# Output file name
output_file = 'h.c1.c2.all.v7.5.1.symbols.gmt'

# Call the function to merge the files
merge_files(filenames, output_file)

# Check if the output file exists
if os.path.exists(output_file):
    print("Files merged successfully!")
else:
    print("File merge failed.")
```
