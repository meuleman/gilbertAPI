```python
import numpy as np
import h5py
import re
import csv
import pandas as pd

##### IMPORT ARCHS4 H5PY DATA #####  

root = "/home/lwu/work/NarrativeGenomics/region_narration/code/pipeline/ARCHS4_LW/"
file_path = "refining/data_slash_results/"
filename = root + file_path + 'archs4_gene_human_v2.1_normalized_LW20230126.h5'

# Open the HDF5 file
with h5py.File(filename, "r") as file:
    # Access the "data" group
    
    # Get the keys (dataset or subgroup names) within the group
    keys = list(file.keys())
    
    # Print the keys
    print("Keys under 'file' group:")
    print()
    for key in keys:
        print(key)

```

```python
# Open the HDF5 file in read-write mode
with h5py.File(filename, "a") as file:
    
    # Access the "data" group
    group = file["data"]
    
    # Check if the objects exist before deleting
    if "highest_tier_category_matrix" in group:
        del group["highest_tier_category_matrix"]
        print("Deleted 'highest_tier_category_matrix'")

    if "lowest_tier_category_matrix" in group:
        del group["lowest_tier_category_matrix"]
        print("Deleted 'lowest_tier_category_matrix'")

```

```python
# Open the HDF5 file in read/write mode
with h5py.File(filename, "r+") as file:
    # Access the "data" group
    group = file["data"]
    
    # Rename the objects
    if "highest_tier_category_matrix2" in group:
        old_dataset = group["highest_tier_category_matrix2"]
        group.create_dataset("highest_tier_category_matrix_qn", data=old_dataset)
        del group["highest_tier_category_matrix2"]
    
    if "lowest_tier_category_matrix2" in group:
        old_dataset = group["lowest_tier_category_matrix2"]
        group.create_dataset("lowest_tier_category_matrix_qn", data=old_dataset)
        del group["lowest_tier_category_matrix2"]
        


```

Groups: 
1. "data" 
    - "raw_matrix"
      The raw (but filtered and catgorized) ARCHS4 v2.1 data
      
    - "threshold_matrix"
      The 5M read count threshold filteration 
      
    - "corrected_matrix"
      Correcting for transcript-length 
      
    - "qn_matrix"
      Performed quantile normalization 
      
    - "highest_tier_category_matrix_raw" 
    - "lowest_tier_category_matrix_raw"
      Raw samples averaged across highest level tier and lowest level tier categories.
    
    - "highest_tier_category_matrix_qn"
    - "lowest_tier_category_matrix_qn"
      Raw samples averaged across highest level tier and lowest level tier categories.
    
    - "high_tier_oe_matrix_raw"
    - "low_tier_oe_matrix_raw"  
      Raw O/E Matrix for highest level and lowest level tier categories.

    - "high_tier_log2oe_matrix_raw"
    - "low_tier_log2oe_matrix_raw"
      Raw log2(O/E) Matrix for highest level and lowest level tier categories.
    
    - "high_tier_oe_matrix_qn"
    - "low_tier_oe_matrix_qn"  
      QN O/E Matrix for highest level and lowest level tier categories.

    - "high_tier_log2oe_matrix_qn"
    - "low_tier_log2oe_matrix_qn"
      QN log2(O/E) Matrix for highest level and lowest level tier categories.
   
     
2. "meta"
i. "biosamples"
    
    - "all_biosample_names"
      Raw matrix - all biosample names
      
    - "threshold_biosample_names"
      Threshold/Corrected/QN matrix - After 5M read count threshold filteration, all biosample names 
    
    - "all_biosample_category"
      Raw matrix - Sample associated category 
      
    - "category"
      QN matrix - Sample associated category 
    
ii. "genes"

    - "all_gene_names"
       All gene names 
       
    - "gene_description"
       Short gene description from gProfiler 
       
    - "gene_id"
      Gene ENSMB ID
      
    - "gene_length"
      Gene transcript length
