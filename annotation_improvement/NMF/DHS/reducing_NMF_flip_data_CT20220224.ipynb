**one at a time saves**

```python
filenames = []
directory_path = '/home/ctrader/Hold/NMF/DHS/fill_in_one_at_a_time_saves/'
directory = os.fsencode(directory_path)
for file in os.listdir(directory):
    filename = os.fsdecode(file)
    if filename[-8:] == '.npy.npz':
        filenames.append(filename)
```

```python
for f in filenames:
    filename = '/home/ctrader/Hold/NMF/DHS/fill_in_one_at_a_time_saves/' + f
    np_file = np.load(filename)
    
    basis = np_file['basis']
    mixture = np_file['mixture']
    
    basis1 = np.float32(basis)
    mixture1 = np.float32(mixture)
    
    new_filename = filename[:-18] + 'CT20220224.npz'
    print(new_filename)
    
    np.savez(new_filename, basis=basis1, mixture=mixture1)
    print(filename)
    print()
    
    os.remove(filename)
```
