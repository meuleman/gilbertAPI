```python
from __future__ import print_function
import os
import sys

##########################################################
## This script combines all the STRING's channels subscores
## into the final combined STRING score.
## It uses unpacked protein.links.full.xx.txt.gz as input
## which can be downloaded from the download subpage:
##      https://string-db.org/cgi/download.pl
##########################################################
 
input_file = "../data/9606.protein.links.full.v10.5.txt"

if not os.path.exists(input_file):
    sys.exit("Can't locate input file %s" % input_file)

prior = 0.041

def compute_prior_away(score, prior):

    if score < prior: score = prior
    score_no_prior = (score - prior) / (1 - prior)

    return score_no_prior

s1_neighborhood = []
s1_neighborhood = []
s1_fusion = []
s1_cooccurrence = []
s1_coexpression = []
s1_coexpression_transferred = []
s1_experiments = []
s1_experiments_transferred = []
s1_database = []
s1_database_transferred = []
s1_textmining = []
s1_textmining_transferred = []

s2_neighborhood = []
s2_neighborhood_transferred = []
s2_fusion = []
s2_cooccurrence = []
s2_coexpression = []
s2_coexpression_transferred = []
s2_experiments = []
s2_experiments_transferred = []
s2_database = []
s2_database_transferred = []
s2_textmining = []
s2_textmining_transferred = []


protein1_list = []
protein2_list = []

# Columns = Product
step4 = []

# Columns = 1-Product with prior multiplication
step5 = []


header = True
for line in open(input_file):

    if header:
        header = False
        continue
    
    l = line.split()
    
    ## load the line
        
    (protein1, protein2,
     neighborhood, neighborhood_transferred,
     fusion, cooccurrence,
     homology,
     coexpression, coexpression_transferred,
     experiments, experiments_transferred,
     database, database_transferred,
     textmining, textmining_transferred,
     initial_combined) = l


    ## STEP 1: divide by 1000

    neighborhood = float(neighborhood) / 1000
    neighborhood_transferred = float(neighborhood_transferred) / 1000
    fusion = float(fusion) / 1000
    cooccurrence =  float(cooccurrence) / 1000
    homology = float(homology) / 1000
    coexpression = float(coexpression) / 1000
    coexpression_transferred = float(coexpression_transferred) / 1000
    experiments = float(experiments) / 1000
    experiments_transferred = float(experiments_transferred) / 1000
    database = float(database) / 1000
    database_transferred = float(database_transferred) / 1000
    textmining = float(textmining) / 1000
    textmining_transferred = float(textmining_transferred) / 1000
    initial_combined = int(initial_combined)

    
    s1_neighborhood.append(neighborhood)
    s1_neighborhood_transferred.append(neighborhood_transferred)
    s1_fusion.append(fusion)
    s1_cooccurrence.append(cooccurrence)
    s1_coexpression.append(coexpression)
    s1_coexpression_transferred.append(coexpression_transferred)
    s1_experiments.append(experiments)
    s1_experiments_transferred.append(experiments_transferred)
    s1_database.append(database)
    s1_database_transferred.append(database_transferred)
    s1_textmining.append(textmining)
    s1_textmining_transferred.append(textmining_transferred)
    
    
    

    ## STEP 2: compute prior away

    neighborhood_prior_corrected                 = compute_prior_away (neighborhood, prior)             
    neighborhood_transferred_prior_corrected     = compute_prior_away (neighborhood_transferred, prior) 
    fusion_prior_corrected                       = compute_prior_away (fusion, prior)             
    cooccurrence_prior_corrected                 = compute_prior_away (cooccurrence, prior)           
    coexpression_prior_corrected                 = compute_prior_away (coexpression, prior)            
    coexpression_transferred_prior_corrected     = compute_prior_away (coexpression_transferred, prior) 
    experiments_prior_corrected                  = compute_prior_away (experiments, prior)   
    experiments_transferred_prior_corrected      = compute_prior_away (experiments_transferred, prior) 
    database_prior_corrected                     = compute_prior_away (database, prior)      
    database_transferred_prior_corrected         = compute_prior_away (database_transferred, prior)
    textmining_prior_corrected                   = compute_prior_away (textmining, prior)            
    textmining_transferred_prior_corrected       = compute_prior_away (textmining_transferred, prior) 
    
    s2_neighborhood.append(neighborhood_prior_corrected)
    s2_neighborhood_transferred.append(neighborhood_transferred_prior_corrected)
    s2_fusion.append(fusion_prior_corrected)
    s2_cooccurrence.append(cooccurrence_prior_corrected)
    s2_coexpression.append(coexpression_prior_corrected)
    s2_coexpression_transferred.append(coexpression_transferred_prior_corrected)
    s2_experiments.append(experiments_prior_corrected)
    s2_experiments_transferred.append(experiments_transferred_prior_corrected)
    s2_database.append(database_prior_corrected)
    s2_database_transferred.append(database_transferred_prior_corrected)
    s2_textmining.append(textmining_prior_corrected)
    s2_textmining_transferred.append(textmining_transferred_prior_corrected)

    ## STEP 3: Correction
    ## then, combine the direct and transferred scores for each category:

    neighborhood_both_prior_corrected = 1.0 - (1.0 - neighborhood_prior_corrected) * (1.0 - neighborhood_transferred_prior_corrected)
    coexpression_both_prior_corrected = 1.0 - (1.0 - coexpression_prior_corrected) * (1.0 - coexpression_transferred_prior_corrected)
    experiments_both_prior_corrected = 1.0 - (1.0 - experiments_prior_corrected) * (1.0 - experiments_transferred_prior_corrected)
    database_both_prior_corrected = 1.0 - (1.0 - database_prior_corrected) * (1.0 - database_transferred_prior_corrected)
    textmining_both_prior_corrected = 1.0 - (1.0 - textmining_prior_corrected) * (1.0 - textmining_transferred_prior_corrected)

    ## now, do the homology correction on cooccurrence and textmining:

    cooccurrence_prior_homology_corrected = cooccurrence_prior_corrected * (1.0 - homology)
    textmining_both_prior_homology_corrected = textmining_both_prior_corrected * (1.0 - homology)
    
    
    s3_neighborhood_both.append(neighborhood_both_prior_corrected)
    s3_coexpression_both.append(coexpression_both_prior_corrected)
    s3_experiments_both.append(experiments_both_prior_corrected)
    s3_database_both.append(database_both_prior_corrected)
    s3_textmining_both.append(textmining_both_prior_corrected)
    s3_cooccurrence_homology.append(cooccurrence_prior_homology_corrected)
    s3_textmining_both_homology.append(textmining_both_prior_homology_corrected)
    
    

    ## STEP 4: next, do the 1 - multiplication:

    combined_score_one_minus = (
        (1.0 - neighborhood_both_prior_corrected) *
        (1.0 - fusion_prior_corrected) *
        (1.0 - cooccurrence_prior_homology_corrected) *
        (1.0 - coexpression_both_prior_corrected) *
        (1.0 - experiments_both_prior_corrected) *
        (1.0 - database_both_prior_corrected) *
        (1.0 - textmining_both_prior_homology_corrected) ) 
    
    step4.append(combined_score_one_minus)

    ## STEP 5: and lastly, do the 1 - conversion again, and put back the prior *exactly once*

    combined_score = (1.0 - combined_score_one_minus)            ## 1- conversion
    combined_score *= (1.0 - prior)                              ## scale down
    combined_score += prior                                      ## and add prior.

    step5.append(combined_score)
    
    ## round

    combined_score = int(combined_score * 1000)
    print(protein1, protein2, combined_score)
    protein1_list.append(protein1)
    protein2_list.append(protein2)
    

step1 = pd.DataFrame([protein1_list, protein2_list, s1_neighborhood, s1_neighborhood_transferred, s1_fusion, 
                      s1_cooccurrence, s1_coexpression, s1_coexpression_transferred, s1_experiments, 
                      s1_experiments_transferred, s1_database, s1_database_transferred, s1_textmining, 
                      s1_textmining_transferred], 
                     columns = ['protein1', 'protein2', 'neighborhood', 'neighborhood_transferred', 'fusion', 
                                'cooccurrence', 'coexpression', 'coexpression_transferred', 'experiments', 
                                'experiments_transferred', 'database', 'database_transferred', 'textmining', 
                                'textmining_transferred']) 
step2 = pd.DataFrame([protein1_list, protein2_list, s2_neighborhood, s2_neighborhood_transferred, s2_fusion, s2_cooccurrence, s2_coexpression, 
                      s2_coexpression_transferred, s2_experiments, s2_experiments_transferred, s2_database, 
                      s2_database_transferred, s2_textmining, s2_textmining_transferred], 
                     columns = ['protein1', 'protein2', 'neighborhood', 'neighborhood_transferred', 'fusion', 
                                'cooccurrence', 'coexpression', 'coexpression_transferred', 'experiments', 
                                'experiments_transferred', 'database', 'database_transferred', 'textmining', 
                                'textmining_transferred'])
step3 = pd.DataFrame([protein1_list, protein2_list, s3_neighborhood_both, s3_coexpression_both, 
                      s3_experiments_both, s3_database_both, s3_textmining_both, s3_cooccurrence_homology, 
                      s3_textmining_both_homology], 
                     columns = ['protein1', 'protein2', 'neighborhood_both', 'coexpression_both', 
                                'experiments_both', 'database_both', 'textmining_both', 'cooccurrence_homology', 
                                'textmining_both_homology'])
```
