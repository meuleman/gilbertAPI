```python
import sys
import numpy as np
import pandas as pd
import gzip
import matplotlib.pyplot as plt
from sklearn.decomposition import NMF, non_negative_factorization
```

```python
sys.path.append('/home/ctrader/vocab/Vocabulary/')
import OONMFhelpers
import OONMF
```

```python
#read in 733 sample x 3.5+ million DHS presence/absence matrix
dhs = pd.read_table('/home/ctrader/Hold/DHS/dat_bin_FDR01_hg38.txt.gz', header=None).T
dhs.shape
```

**Immune**

```python
#Fetch the new-to-be-added immune data, mapped to the 733-biosample Index
dat_immune = pd.read_table('/home/ctrader/vocab/Vocabulary/data/dat_immunopedia.txt.gz', header=None).T
dat_immune.shape
```

```python
#Combine the two datasets
dhs_immune = np.concatenate([dhs, dat_immune])
dhs_immune.shape
```

```python
#Load in NMF decomposition based on 733 biosamples
fn = gzip.GzipFile('/home/ctrader/Hold/DHS/2018-06-08NC16_NNDSVD_Mixture.npy.gz', "r")
decomp_Mixture = np.load(fn)
decomp_Mixture.shape
```

```python
#Solve the model for a fixed matrix H
W, H, n_iter = non_negative_factorization(dhs_immune, n_components=16, init='custom', random_state=3, update_H=False, H=decomp_Mixture, verbose=1)
```

```python
W.shape
```

```python
f = gzip.GzipFile('/home/ctrader/AG/AugmentedGenomics/recommendersystems/NMF/DHS/2020-04-14ImmuneSpecialNC16seed20.npy.gz', "w")
np.save(f, W)
f.close()
```
