```python
#Example code from LightFM

from lightfm import LightFM
from lightfm.datasets import fetch_movielens
from lightfm.evaluation import precision_at_k
from lightfm.evaluation import recall_at_k

# Load the MovieLens 100k dataset. Only five
# star ratings are treated as positive.
data_movie = fetch_movielens(min_rating=5.0)

# Instantiate and train the model
model_movie = LightFM(loss='warp')
model_movie.fit(data_movie['train'], epochs=30, num_threads=2)

# Evaluate the trained model
test_precision = precision_at_k(model_movie, data_movie['test'], k=10).mean()
print(test_precision)
test_recall = recall_at_k(model_movie, data_movie['test'], k=10).mean()
print(test_recall)
```
